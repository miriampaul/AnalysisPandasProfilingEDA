```python
import pandas as pd
dataset=pd.read_csv("/content/ridge.csv")
df=dataset.copy()
#Split into X (Independent) and Y(Predicted)

X=df.iloc[:,:-1]
Y=df.iloc[:,-1]
```

```python
X
```

```python
#Ridge Regression
from sklearn.linear_model import Ridge
```

```python
l=1
ridge=Ridge(alpha=l)
```

```python
ridge.fit(X, Y)
```

```python
#get the coefficient
ridge_coeff=ridge.coef_
```

```python
#get the intercept
ridge_intercept=ridge.intercept_
```

```python
#OLS Y=2*x+0
#Alpha as 1 Y=1.667*x+0.883

#Plot the ridge
X_plt=[0,1 ,2 ,3, 4]
Y_plt=ridge.predict(pd.DataFrame(X_plt))
```

```python
import matplotlib.pyplot as plt
plt.figure(1)
plt.plot(X_plt, Y_plt)
plt.ylim(ymin=0, ymax=9)
plt.xlim(xmin=0, xmax=6)
#Add the equation text to the line
plt.text(X_plt[-1] + 0.1, Y_plt[-1], " y = "+str("%.2f" %ridge_coeff[0])+ " * x + " + str("%.2f"%ridge_intercept)+ "for \u03BB or \u03B1"+ str(1))
```
