```python
# import the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#import datasets
dataset=pd.read_csv("/content/Data.csv")
#Features (1st few columns) and #dependent columns (last columns)
X=dataset.iloc[:,:-1].values # all columns excluding the last column
Y=dataset.iloc[:,-1].values #include only last columns
Y
```

```python
dataset
```

```python
print(X)
```

```python
print(Y)
```

```python
#Taking Care of Missing Data
from sklearn.impute import SimpleImputer
imputer= SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, 1:3])
X[:, 1:3]=imputer.transform(X[:,1:3])
```

```python
print(X)
```

```python
#Encoding the categorical data
```

```python
#Encoding the independent variable
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct= ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])] , remainder='passthrough' )
X=np.array(ct.fit_transform(X))
X
```

```python
#Encoding the Dependent Variable
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
Y=le.fit_transform(Y)
```

```python
Y
```

```python
#Splitting the dataset into Training and Test Set
```

```python
 from sklearn.model_selection import train_test_split
 X_train, X_test, Y_train, Y_test=\
 train_test_split(X, Y , test_size=0.2, random_state=1,stratify=Y)
```
