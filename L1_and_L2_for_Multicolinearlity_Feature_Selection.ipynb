```python
import pandas as pd
dataset=pd.read_csv('mcl.csv')
df=dataset.copy()
#Split the X(Independent) and Y(predicted)
X=df.iloc[:, :-1]
Y=df.iloc[:,-1]
```

```python
#Correlaton matrix of the dataset
correlation=X.corr()
```

```python
correlation
```

```python
#perform the three refressions on the data
from sklearn.linear_model import Lasso, Ridge, LinearRegression
#Perform the Linear Regression

lr=LinearRegression()
lr.fit(X, Y)
```

```python
#Get the coefficient and intercept
```

```python
lr_coeff=lr.coef_
lr_intercept=lr.intercept_
```

```python
lr_coeff
```

```python
#lasso regression
lasso=Lasso(alpha=10)
lasso.fit(X, Y)
lasso_coeff=lasso.coef_
lasso_intercept=lasso.intercept_
```

```python
#Ridge regression
ridge=Ridge(alpha=100)
ridge.fit(X, Y)
ridge_coeff=ridge.coef_
ridge_intercept=ridge.intercept_
```

```python
 lr_coeff
 lasso_coeff
 ridge_coeff
```
