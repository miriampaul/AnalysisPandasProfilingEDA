```python
#Implement and see the effect of Lasso Regression
#Using plots of different Lasso Regression Lines
#Import pandas for Data processing
import pandas as pd
dataset=pd.read_csv("/content/ridge.csv")
```

```python
df=dataset.copy()
X=df.iloc[:,:-1]
Y=df.iloc[:, -1]
```

```python
#import Lasso and matplotlib
from sklearn.linear_model import Lasso
```

```python
import matplotlib.pyplot as plt
```

```python
#Create test data for plotting
X_plt=[0, 1, 2, 3, 4]
```

```python
#Create a list of different alpha/penatly parameter values
lasso_l=[0, 0.5, 1, 2, 4]
```

```python
#Plot the different lasso regression lines in one figures
for i, l in enumerate(lasso_l):
  lasso=Lasso(alpha=l)
  lasso.fit(X, Y)
  lasso_coeff=lasso.coef_
  lasso_intercept=lasso.intercept_
  Y_plt=lasso.predict(pd.DataFrame(X_plt))
  plt.figure(1)
  plt.plot(X_plt, Y_plt)
  plt.ylim(ymin=0, ymax=9)
  plt.xlim(xmin=0, xmax=6)
  plt.text(X_plt[-1], Y_plt[-1],
           'y= '+
           str("%.2f" %lasso_coeff)+
           " * x "+
           " + "+
           str("%.2f" %lasso_intercept)+
           "  for \u033B or \u03B1"+str(l))
```
